```python
import pandas as pd
import requests 
import os
from IPython.display import Audio, display

df = pd.read_csv("../data/SpotGenTrack/Data Sources/spotify_tracks.csv")
print(df.columns.tolist())
```

```python
sample = df[df["preview_url"].notna()].sample(3, random_state=0)
for url in sample["preview_url"]:
    response = requests.get(url)
    response.raise_for_status()
    display(Audio(response.content, autoplay = False))
```
